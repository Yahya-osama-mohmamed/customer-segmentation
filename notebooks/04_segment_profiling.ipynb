# 04 — Segment Profiling
Clusters become **business segments**: named from their R/F/M profile so a
marketing stakeholder can read the table without knowing what a centroid is.

```python
import sys, os
sys.path.append(os.path.abspath(".."))
import pandas as pd
import matplotlib.pyplot as plt
plt.style.use("seaborn-v0_8-whitegrid")

prof = pd.read_csv("../data/processed/segment_profile.csv", index_col=0)
prof
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 4.6))
axes[0].pie(prof["Customers"], labels=prof.index, autopct="%1.0f%%",
            pctdistance=0.8)
axes[0].set_title("Customers by segment")
axes[1].bar(prof.index, prof["RevenueShare"], color="#B85C38")
axes[1].set_title("Revenue share by segment (%)")
axes[1].tick_params(axis="x", rotation=20)
fig.tight_layout(); fig.savefig("figures/segment_profile.png", dpi=150)
plt.show()
```

**Reading the table** — the punchline of the whole project is usually that a
small "Champions" segment produces the majority of revenue, while a large
"Hibernating" tail produces almost none: retention budget belongs on the
former, cheap reactivation campaigns (or nothing) on the latter. "At Risk"
customers — lapsed but historically valuable — are the classic win-back
target and the most actionable list this analysis produces.
